```python
# ============================================================
# 8. ADIM: MODEL VE VERİ SETİ KONTROLÜ
# Standart LoRA kullanacağız; kuantizasyon uygulamayacağız.
# ============================================================

import os
import json
from transformers import AutoTokenizer

# Kullanacağımız 3 milyar parametreli temel model.
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# Proje klasörünün ve veri dosyasının yolları.
PROJECT_DIR = "/content/drive/MyDrive/LoRA_Pilot_CodeCarbon_Yeni"
DATASET_PATH = os.path.join(PROJECT_DIR, "pilot_dataset.json")

# Kaydettiğimiz veri setini Drive'dan okuyoruz.
with open(DATASET_PATH, "r", encoding="utf-8") as f:
    dataset = json.load(f)

# Veri setinde tam olarak 50 örnek olduğunu doğruluyoruz.
assert len(dataset) == 50, f"50 örnek bekleniyordu, bulunan: {len(dataset)}"

# Modelin tokenizer'ını yüklüyoruz.
# Bu işlem modelin bütün ağırlıklarını GPU'ya yüklemez.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# İlk soru-cevap örneğini eğitimde kullanılacak metne dönüştürüyoruz.
first_example = dataset[0]
formatted_text = (
    "### Soru:\n"
    + first_example["instruction"]
    + "\n\n### Cevap:\n"
    + first_example["response"]
)

# Örneği tokenlara ayırıyoruz ve token sayısını kontrol ediyoruz.
tokens = tokenizer(formatted_text, add_special_tokens=True)

# Kontrol sonuçlarını yazdırıyoruz.
print("Temel model:", MODEL_NAME)
print("Yöntem: Standart LoRA")
print("Kuantizasyon: Yok")
print("Veri setindeki örnek sayısı:", len(dataset))
print("Tokenizer başarıyla yüklendi.")
print("İlk örneğin token sayısı:", len(tokens["input_ids"]))
print("\nİlk örnek:")
print(formatted_text)
```

```python
# ============================================================
# 9. ADIM: QWEN2.5-3B MODELİNİ FP16 İLE YÜKLEME
# Standart LoRA kullanacağız; kuantizasyon yok.
# ============================================================

import torch
from transformers import AutoModelForCausalLM

# ------------------------------------------------------------
# 1. Model kimliğini belirliyoruz.
# Tokenizer ile aynı model kullanılmalı.
# ------------------------------------------------------------
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# ------------------------------------------------------------
# 2. GPU'nun kullanılabilir olduğunu kontrol ediyoruz.
# GPU yoksa model yüklemeyi başlatmıyoruz.
# ------------------------------------------------------------
assert torch.cuda.is_available(), (
    "GPU bulunamadı. Colab çalışma zamanı ayarlarından GPU seç."
)

# ------------------------------------------------------------
# 3. Modeli FP16 biçiminde yüklüyoruz.
# torch.float16, ağırlıkları 16 bitlik kayan nokta biçiminde tutar.
# device_map='cuda:0', modeli ilk GPU'ya yerleştirir.
# load_in_4bit veya load_in_8bit kullanmadığımız için
# burada kuantizasyon uygulamıyoruz.
# ------------------------------------------------------------
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="cuda:0",
    low_cpu_mem_usage=True
)

# ------------------------------------------------------------
# 4. Modelin kuantizasyon kullanmadan yüklendiğini kontrol ediyoruz.
# ------------------------------------------------------------
print("Model başarıyla yüklendi.")
print("Model:", MODEL_NAME)
print("Yöntem: Standart LoRA için temel model")
print("Kuantizasyon: Yok")
print("Model ağırlıklarının veri tipi:", model.dtype)
print("Modelin bulunduğu cihaz:", model.device)

# ------------------------------------------------------------
# 5. GPU belleğini ölçüyoruz.
# Bu değer, PyTorch'un ayırdığı belleği gösterir.
# GPU belleğinin tamamının yalnızca bu model tarafından
# kullanıldığını ifade etmez.
# ------------------------------------------------------------
allocated_gib = torch.cuda.memory_allocated(0) / (1024 ** 3)
reserved_gib = torch.cuda.memory_reserved(0) / (1024 ** 3)

print(f"PyTorch tarafından ayrılan GPU belleği: {allocated_gib:.2f} GiB")
print(f"PyTorch tarafından rezerve edilen GPU belleği: {reserved_gib:.2f} GiB")

# ------------------------------------------------------------
# 6. GPU'nun toplam ve boş belleğini görüntülüyoruz.
# Bu kontrol, eğitim öncesinde ne kadar alan kaldığını
# değerlendirmemize yardımcı olur.
# ------------------------------------------------------------
gpu_properties = torch.cuda.get_device_properties(0)
total_gib = gpu_properties.total_memory / (1024 ** 3)
free_bytes, total_bytes = torch.cuda.mem_get_info(0)
free_gib = free_bytes / (1024 ** 3)

print(f"GPU toplam belleği: {total_gib:.2f} GiB")
print(f"Şu anda boş GPU belleği: {free_gib:.2f} GiB")
```

```python

# ============================================================
# 10. ADIM: STANDART LoRA YAPILANDIRMASINI MODELE UYGULAMA
# ============================================================

# Gerekli sınıf ve fonksiyonu PEFT kütüphanesinden içe aktarıyoruz.
# PEFT, büyük dil modellerini tüm parametrelerini eğitmeden uyarlamamızı sağlar.
from peft import LoraConfig, get_peft_model

# LoRA ayarlarını belirliyoruz.
lora_config = LoraConfig(
    # LoRA matrislerinin rank değeridir.
    # Pilot deneyimizde r=8 kullanıyoruz.
    r=8,

    # LoRA güncellemesinin ölçeğini belirler.
    # Burada alpha=16 ve rank=8 seçilmiştir.
    lora_alpha=16,

    # Eğitim sırasında LoRA katmanlarında uygulanacak dropout oranıdır.
    # Aşırı öğrenmeyi azaltmaya yardımcı olabilir.
    lora_dropout=0.05,

    # LoRA uygulanacak dikkat (attention) projeksiyonlarını belirtiyoruz.
    # q_proj: Query projeksiyonu
    # v_proj: Value projeksiyonu
    # Bu pilotta yalnızca Q ve V katmanlarını hedefliyoruz.
    target_modules=["q_proj", "v_proj"],

    # Temel modelin bias parametrelerini LoRA kapsamında eğitmiyoruz.
    bias="none",

    # Modelin nedensel dil modeli (causal language modeling) görevinde
    # kullanılacağını belirtiyoruz.
    task_type="CAUSAL_LM"
)

# LoRA adaptörlerini temel modele ekliyoruz.
# Temel modelin ağırlıkları dondurulur; LoRA parametreleri eğitilebilir.
model = get_peft_model(model, lora_config)

# Eğitilebilir ve toplam parametre sayılarını görüntülüyoruz.
# Böylece modelin ne kadar küçük bir bölümünün eğitileceğini görebiliriz.
model.print_trainable_parameters()

# LoRA yapılandırmasının beklediğimiz gibi uygulandığını kontrol ediyoruz.
print("\nLoRA yapılandırması başarıyla uygulandı.")
print("Hedef modüller:", lora_config.target_modules)
print("LoRA rank (r):", lora_config.r)
print("LoRA alpha:", lora_config.lora_alpha)
print("LoRA dropout:", lora_config.lora_dropout)

# Temel modelin ağırlıklarının FP16 biçiminde kaldığını kontrol ediyoruz.
# Bu, kuantizasyon kullanmadığımız standart LoRA deneyidir.
print(
    "Model parametre veri türü:",
    next(model.parameters()).dtype
)

# Modelin GPU üzerinde olup olmadığını kontrol ediyoruz.
print(
    "Modelin bulunduğu cihaz:",
    next(model.parameters()).device
)
```

```python
# ============================================================
# 10.5. ADIM: QWEN2.5-3B-INSTRUCT TEMEL MODELİNİ YÜKLEME
# ============================================================

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Kullanacağımız temel modelin Hugging Face üzerindeki adını belirliyoruz.
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# Modelin bellekte FP16 (16-bit kayan nokta) biçiminde tutulmasını
# sağlıyoruz. Bu, FP32'ye kıyasla model ağırlıklarının bellekte
# daha az yer kaplamasına yardımcı olur.
MODEL_DTYPE = torch.float16

# GPU'nun kullanılabilir olduğunu doğruluyoruz.
assert torch.cuda.is_available(), "GPU bulunamadı; çalışma durduruldu."

# Tokenizer'ı yüklüyoruz.
# Tokenizer, metinleri modelin işleyebileceği token kimliklerine dönüştürür.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Veri hazırlığı ve eğitim sırasında padding gerekebileceği için
# tokenizer'da padding token'ı tanımlı değilse EOS token'ını kullanıyoruz.
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Temel modeli GPU'nun ilk cihazına yüklüyoruz.
# Bu, standart LoRA deneyidir; 4-bit kuantizasyon kullanmıyoruz.
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=MODEL_DTYPE,
    device_map="cuda:0",
    low_cpu_mem_usage=True
)

# Eğitim sırasında bellek kullanımını azaltmaya yardımcı olması için
# gradient checkpointing'i daha sonra etkinleştireceğiz.
# Şimdilik yalnızca modelin doğru yüklendiğini kontrol ediyoruz.

print("\nTemel model başarıyla yüklendi.")
print("Model:", MODEL_NAME)
print("Veri türü:", next(model.parameters()).dtype)
print("Model cihazı:", next(model.parameters()).device)
print("GPU:", torch.cuda.get_device_name(0))

# GPU belleği kullanımını GiB cinsinden görüntülüyoruz.
allocated_gib = torch.cuda.memory_allocated(0) / (1024 ** 3)
reserved_gib = torch.cuda.memory_reserved(0) / (1024 ** 3)

print(f"GPU belleği ayrılmış: {allocated_gib:.2f} GiB")
print(f"GPU belleği rezerve edilmiş: {reserved_gib:.2f} GiB")
```

```python

# ============================================================
# 10.6. ADIM: LoRA ADAPTÖRLERİNİ TEMEL MODELE UYGULAMA
# ============================================================

from peft import LoraConfig, get_peft_model

# LoRA yapılandırmasını oluşturuyoruz.
lora_config = LoraConfig(
    # LoRA'nın düşük rank değerini belirliyoruz.
    # Pilot deneyinde r=8 kullanıyoruz.
    r=8,

    # LoRA güncellemesinin ölçeğini belirliyoruz.
    lora_alpha=16,

    # Eğitim sırasında LoRA katmanlarında kullanılacak dropout oranı.
    lora_dropout=0.05,

    # Yalnızca Query ve Value projeksiyonlarına LoRA ekliyoruz.
    # q_proj: Query projeksiyonu
    # v_proj: Value projeksiyonu
    target_modules=["q_proj", "v_proj"],

    # Temel modelin bias parametrelerini eğitmiyoruz.
    bias="none",

    # Modelin nedensel dil modeli görevi için kullanılacağını belirtiyoruz.
    task_type="CAUSAL_LM"
)

# LoRA adaptörlerini modele ekliyoruz.
# Temel modelin ağırlıkları dondurulur; LoRA parametreleri eğitilebilir.
model = get_peft_model(model, lora_config)

# Eğitilebilir ve toplam parametre sayılarını görüntülüyoruz.
model.print_trainable_parameters()

# LoRA ayarlarını kontrol ediyoruz.
print("\n--- LoRA yapılandırma kontrolü ---")
print("Hedef modüller:", lora_config.target_modules)
print("Rank (r):", lora_config.r)
print("Alpha:", lora_config.lora_alpha)
print("Dropout:", lora_config.lora_dropout)

# Modelin hâlâ GPU üzerinde olduğunu kontrol ediyoruz.
print("\nModel cihazı:", next(model.parameters()).device)
print("GPU:", torch.cuda.get_device_name(0))

# LoRA parametrelerinin eğitilebilir olup olmadığını kontrol ediyoruz.
trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)
total_params = sum(p.numel() for p in model.parameters())

print("Eğitilebilir parametre sayısı:", f"{trainable_params:,}")
print("Toplam parametre sayısı:", f"{total_params:,}")
print(
    "Eğitilebilir parametre yüzdesi:",
    f"{100 * trainable_params / total_params:.4f}%"
)
```

```python
# ============================================================
# 10.7. ADIM: EĞİTİM ÖNCESİ BELLEK VE MODEL AYARLARI
# ============================================================

import torch

# Modelin eğitim sırasında KV cache kullanmasını kapatıyoruz.
# Eğitimde KV cache genellikle gerekli değildir ve ek bellek tüketebilir.
model.config.use_cache = False

# Gradient checkpointing'i etkinleştiriyoruz.
# Bu yöntem bazı ara aktivasyonları yeniden hesaplayarak
# eğitim sırasında gereken GPU belleğini azaltmaya yardımcı olur.
model.gradient_checkpointing_enable()

# LoRA parametrelerinin eğitilebilir olduğunu doğruluyoruz.
trainable_params = [
    name for name, param in model.named_parameters()
    if param.requires_grad
]

assert len(trainable_params) > 0, (
    "Hata: Eğitilebilir LoRA parametresi bulunamadı."
)

# Temel modelin ağırlıklarının FP16 olduğunu kontrol ediyoruz.
print("Model veri türü:", next(model.parameters()).dtype)

# GPU belleği kullanımını ölçüyoruz.
allocated_gib = torch.cuda.memory_allocated(0) / (1024 ** 3)
reserved_gib = torch.cuda.memory_reserved(0) / (1024 ** 3)

print("Eğitilebilir parametre sayısı:", len(trainable_params))
print(f"GPU belleği ayrılmış: {allocated_gib:.2f} GiB")
print(f"GPU belleği rezerve edilmiş: {reserved_gib:.2f} GiB")

print("KV cache:", model.config.use_cache)
print("Gradient checkpointing etkinleştirildi.")
print("Eğitim öncesi ayarlar tamamlandı; eğitim henüz başlamadı.")
```

```python
# Eğitilebilir parametre tensörlerini değil, bu tensörlerdeki
# tüm sayısal parametreleri topluyoruz.

trainable_parameter_count = sum(
    param.numel()
    for param in model.parameters()
    if param.requires_grad
)

# Modeldeki tüm sayısal parametreleri topluyoruz.
total_parameter_count = sum(
    param.numel()
    for param in model.parameters()
)

# Sonuçları ekrana yazdırıyoruz.
print("Eğitilebilir sayısal parametre toplamı:",
      f"{trainable_parameter_count:,}")

print("Modeldeki toplam sayısal parametre:",
      f"{total_parameter_count:,}")

print("Eğitilebilir parametre yüzdesi:",
      f"{100 * trainable_parameter_count / total_parameter_count:.4f}%")
```

```python
# ============================================================
# 10.14. ADIM: TEMEL MODELİ VE TOKENIZER'I YÜKLEME
# ============================================================

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Kullanacağımız temel modeli belirliyoruz.
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# Tokenizer, metinleri modelin işleyebileceği token'lara dönüştürür.
# Modelin kendi tokenizer'ını kullanıyoruz.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Eğitim sırasında padding gerektiğinde kullanılacak token'ı
# tanımlıyoruz. Bu modelde ayrı bir padding token'ı yoksa
# EOS token'ını kullanıyoruz.
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Temel modeli Tesla T4 GPU'ya FP16 hassasiyetinde yüklüyoruz.
# Bu işlem standart LoRA içindir; 4-bit kuantizasyon yapmaz.
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=torch.float16,
    device_map="cuda:0",
    low_cpu_mem_usage=True
)

# Padding işlemlerinde aynı token kimliğinin kullanılmasını sağlıyoruz.
model.config.pad_token_id = tokenizer.pad_token_id

# Modelin doğru cihaza ve hassasiyete yüklendiğini kontrol ediyoruz.
print("Model adı:", MODEL_NAME)
print("Tokenizer başarıyla yüklendi.")
print("Model başarıyla yüklendi.")
print("Model cihazı:", model.device)
print("Model veri tipi:", model.dtype)
print("GPU:", torch.cuda.get_device_name(0))
print(
    "GPU belleği (ayrılmış):",
    round(torch.cuda.memory_allocated(0) / 1024**3, 2),
    "GiB"
)

print("\nModel yükleme kontrolü tamamlandı. Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.15. ADIM: MODELE STANDART LoRA KATMANLARINI EKLEME
# ============================================================

from peft import LoraConfig, get_peft_model

# LoRA yönteminin temel ayarlarını belirliyoruz.
lora_config = LoraConfig(
    r=8,                              # LoRA'nın rank değeri
    lora_alpha=16,                    # LoRA güncellemesinin ölçekleme katsayısı
    lora_dropout=0.05,                # Eğitim sırasında kullanılacak dropout oranı
    target_modules=["q_proj", "v_proj"],  # LoRA uygulanacak dikkat katmanları
    bias="none",                      # Bias parametrelerini eğitmeyeceğiz
    task_type="CAUSAL_LM"              # Metin üretimi görevi
)

# LoRA katmanlarını temel modele ekliyoruz.
# Temel modelin ağırlıkları dondurulur; LoRA parametreleri eğitilebilir.
model = get_peft_model(model, lora_config)

# Eğitilebilir ve toplam parametre sayılarını kontrol ediyoruz.
model.print_trainable_parameters()

# Eğitim için GPU belleği kullanımını görüntülüyoruz.
print("\nGPU:", torch.cuda.get_device_name(0))
print(
    "GPU belleği (ayrılmış):",
    round(torch.cuda.memory_allocated(0) / 1024**3, 2),
    "GiB"
)

print("\nStandart LoRA katmanları eklendi. Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.20. ADIM: MODELİ YÜKLEME VE STANDART LoRA EKLEME
# ============================================================

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import LoraConfig, get_peft_model

# Kullanacağımız temel modeli belirliyoruz.
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# Tokenizer'ı yüklüyoruz.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Padding için ayrı token yoksa EOS token'ını kullanıyoruz.
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Temel modeli Tesla T4 GPU'ya FP16 hassasiyetinde yüklüyoruz.
# Bu işlemde 4-bit kuantizasyon uygulanmıyor.
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=torch.float16,
    device_map="cuda:0",
    low_cpu_mem_usage=True
)

# Modelin padding ayarını yapıyoruz.
model.config.pad_token_id = tokenizer.pad_token_id

print("Temel model başarıyla yüklendi.")

# Standart LoRA yapılandırmasını oluşturuyoruz.
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none",
    task_type="CAUSAL_LM"
)

# LoRA katmanlarını modele ekliyoruz.
model = get_peft_model(model, lora_config)

# Kaç parametrenin eğitilebilir olduğunu gösteriyoruz.
model.print_trainable_parameters()

# GPU belleğini kontrol ediyoruz.
print("\nGPU:", torch.cuda.get_device_name(0))
print(
    "GPU belleği (ayrılmış):",
    round(torch.cuda.memory_allocated(0) / 1024**3, 2),
    "GiB"
)

print("\nModel ve LoRA hazırlığı tamamlandı. Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.21. ADIM: EĞİTİM ÖNCESİ MODEL AYARLARI
# ============================================================

# Eğitim sırasında KV cache kullanmayacağız.
# Bu ayar, eğitim için gereksiz bellek kullanımını önlemeye yardımcı olur.
model.config.use_cache = False

# Gradient checkpointing, ara hesaplamaların bir kısmını yeniden
# hesaplayarak eğitim sırasında bellek ihtiyacını azaltmaya yardımcı olur.
model.gradient_checkpointing_enable()

# LoRA katmanlarının hâlâ eğitilebilir olduğunu kontrol ediyoruz.
trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

total_params = sum(p.numel() for p in model.parameters())

print("KV cache:", model.config.use_cache)
print("Gradient checkpointing:", model.is_gradient_checkpointing)
print("Eğitilebilir parametre sayısı:", trainable_params)
print("Toplam parametre sayısı:", total_params)
print("GPU:", torch.cuda.get_device_name(0))
print(
    "GPU belleği (ayrılmış):",
    round(torch.cuda.memory_allocated(0) / 1024**3, 2),
    "GiB"
)

print("\nEğitim öncesi ayarlar tamamlandı. Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.22. ADIM: VERİ SETİNİ EĞİTİM FORMATINA DÖNÜŞTÜRME
# ============================================================

import os
import json

# Drive'daki veri setini okuyoruz.
PROJECT_DIR = "/content/drive/MyDrive/LoRA_Pilot_CodeCarbon_Yeni"
DATASET_PATH = os.path.join(PROJECT_DIR, "pilot_dataset.json")

with open(DATASET_PATH, "r", encoding="utf-8") as f:
    dataset = json.load(f)

# Her soruyu ve cevabı tek bir eğitim metninde birleştiriyoruz.
# Böylece model, sorunun ardından ilgili cevabı öğrenebilecek.
formatted_dataset = []

for sample in dataset:
    text = (
        "### Soru:\n"
        + sample["instruction"]
        + "\n\n### Cevap:\n"
        + sample["response"]
    )

    formatted_dataset.append({
        "id": sample["id"],
        "text": text
    })

# Veri sayısını ve biçimini doğruluyoruz.
assert len(formatted_dataset) == 50
assert all(item["text"].strip() for item in formatted_dataset)

print("Kaynak örnek sayısı:", len(dataset))
print("Hazırlanan örnek sayısı:", len(formatted_dataset))

print("\n--- İlk örneğin eğitim metni ---")
print(formatted_dataset[0]["text"])

print("\nİlk örnek ID:", formatted_dataset[0]["id"])
print("Son örnek ID:", formatted_dataset[-1]["id"])
print("\nVeri formatı kontrol edildi. Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.23. ADIM: EĞİTİM BİLEŞENLERİNİ KONTROL ETME
# ============================================================

# Trainer, eğitim sürecini yöneten Transformers bileşenidir.
# TrainingArguments ise epoch, öğrenme oranı ve batch size gibi
# eğitim ayarlarını tanımlamak için kullanılır.

from transformers import Trainer, TrainingArguments, DataCollatorForLanguageModeling

print("Trainer: kullanılabilir")
print("TrainingArguments: kullanılabilir")
print("DataCollatorForLanguageModeling: kullanılabilir")

# Eğitim metinlerinin tokenizer tarafından işlenebildiğini
# tek bir örnek üzerinde kontrol ediyoruz.
test_encoding = tokenizer(
    formatted_dataset[0]["text"],
    truncation=True,
    max_length=512
)

print("\nİlk örnekteki token sayısı:", len(test_encoding["input_ids"]))
print("Tokenizer kontrolü başarılı.")
print("Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.24. ADIM: EĞİTİM YAPILANDIRMASINI HAZIRLAMA
# ============================================================

from transformers import TrainingArguments, DataCollatorForLanguageModeling

# Eğitim sırasında kullanılacak temel ayarları tanımlıyoruz.
OUTPUT_DIR = os.path.join(PROJECT_DIR, "lora_pilot_output")

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    num_train_epochs=1,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    logging_steps=1,
    save_strategy="epoch",
    fp16=True,
    gradient_checkpointing=True,
    report_to="none",
    seed=42,
    dataloader_pin_memory=True
)

# Causal language modeling için veri birleştiriciyi hazırlıyoruz.
# Padding token'ları kayıp hesabına katılmayacak.
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

print("Eğitim yapılandırması oluşturuldu.")
print("Epoch:", training_args.num_train_epochs)
print("Batch size:", training_args.per_device_train_batch_size)
print("Gradient accumulation:", training_args.gradient_accumulation_steps)
print("Learning rate:", training_args.learning_rate)
print("FP16:", training_args.fp16)
print("Çıktı klasörü:", OUTPUT_DIR)
print("\nHenüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.25. ADIM: EĞİTİM VERİ KÜMESİNİ HAZIRLAMA
# ============================================================

from datasets import Dataset

# Önceden hazırladığımız 50 metni Hugging Face Dataset
# biçimine dönüştürüyoruz.
# Trainer bu biçimdeki veri kümesiyle çalışabilir.
train_dataset = Dataset.from_list(formatted_dataset)

# Metinleri token'lara dönüştürüyoruz.
# max_length=512, tek bir örneğin uzunluğunu en fazla
# 512 token ile sınırlar.
def tokenize_function(example):
    return tokenizer(
        example["text"],
        truncation=True,
        max_length=512
    )

tokenized_train_dataset = train_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["id", "text"]
)

# Veri kümesinin boyutunu ve ilk örneğin token sayısını kontrol ediyoruz.
print("Eğitim örneği sayısı:", len(tokenized_train_dataset))
print(
    "İlk örneğin token sayısı:",
    len(tokenized_train_dataset[0]["input_ids"])
)
print(
    "İlk örneğin alanları:",
    list(tokenized_train_dataset[0].keys())
)

# 50 örneğin tamamının işlendiğini doğruluyoruz.
assert len(tokenized_train_dataset) == 50
assert "input_ids" in tokenized_train_dataset.column_names

print("\nEğitim veri kümesi hazır.")
print("Henüz eğitim başlamadı.")
```

```python
# ============================================================
# 10.26. ADIM: TRAINER NESNESİNİ OLUŞTURMA
# ============================================================

from transformers import Trainer

# Trainer'a modelimizi, eğitim ayarlarımızı ve token'lara
# dönüştürülmüş veri kümemizi veriyoruz.
# Data collator, aynı batch içindeki örnekleri uygun biçimde
# bir araya getirir ve padding işlemini yönetir.
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    data_collator=data_collator
)

# Eğitim başlamadan önce temel ayarları kontrol ediyoruz.
print("Trainer başarıyla oluşturuldu.")
print("Eğitim örneği sayısı:", len(trainer.train_dataset))
print("Epoch sayısı:", trainer.args.num_train_epochs)
print("Batch size:", trainer.args.per_device_train_batch_size)
print("Gradient accumulation:", trainer.args.gradient_accumulation_steps)
print("Çıktı klasörü:", trainer.args.output_dir)

print("\nTrainer kontrolü tamamlandı. Henüz eğitim başlamadı.")
```

```python

# ============================================================
# 11. ADIM: CODECARBON ÖLÇÜMÜ İLE LORA EĞİTİMİ
# ============================================================

import time
import torch
from codecarbon import EmissionsTracker

# GPU belleği ölçümünü yeni eğitim aşaması için sıfırlıyoruz.
torch.cuda.reset_peak_memory_stats()

# Eğitimin enerji ve karbon emisyonunu ölçmek için
# CodeCarbon takipçisini hazırlıyoruz.
tracker = EmissionsTracker(
    project_name="LoRA_Pilot",
    output_dir=PROJECT_DIR,
    log_level="error"
)

print("CodeCarbon ölçümü başlatılıyor...")

# Enerji ölçümünü eğitim başlamadan önce açıyoruz.
tracker.start()

# Gerçek eğitim süresini saniye cinsinden ölçüyoruz.
start_time = time.time()

try:
    # LoRA modelini 50 örnek üzerinde 1 epoch eğitiyoruz.
    train_result = trainer.train()

finally:
    # Eğitim başarılı olsa da hata verse de ölçümü durdurmaya çalışıyoruz.
    training_time_seconds = time.time() - start_time
    emissions_kg = tracker.stop()

# Eğitim boyunca kullanılan en yüksek GPU belleğini GiB cinsinden hesaplıyoruz.
peak_vram_gib = torch.cuda.max_memory_allocated() / (1024 ** 3)

print("\nEĞİTİM AŞAMASI TAMAMLANDI")
print("Eğitim süresi (saniye):", round(training_time_seconds, 2))
print("Tepe GPU belleği (GiB):", round(peak_vram_gib, 3))
print("CodeCarbon emisyon sonucu (kg CO2 eşdeğeri):", emissions_kg)
print("Trainer eğitim metrikleri:", train_result.metrics)
```

```python

# ============================================================
# 13. ADIM: INFERENCE SORUSUNU VERİ SETİYLE KARŞILAŞTIRMA
# ============================================================

# Kontrol edeceğimiz test sorusunu belirliyoruz.
test_question = "Makine öğrenmesi nedir?"

# Veri setindeki soru metinlerini alıyoruz.
dataset_questions = [
    sample["instruction"].strip()
    for sample in dataset
]

# Aynı sorunun birebir bulunup bulunmadığını kontrol ediyoruz.
exact_match = any(
    question.casefold() == test_question.strip().casefold()
    for question in dataset_questions
)

print("Veri setindeki örnek sayısı:", len(dataset_questions))
print("Test sorusu:", test_question)
print("Soru veri setinde birebir mevcut mu?:", exact_match)

print("\nVeri setindeki tüm sorular:")
for i, question in enumerate(dataset_questions, start=1):
    print(f"{i:02d}. {question}")
```

```python

# ============================================================
# 13. ADIM: VERİ SETİNDE BULUNMAYAN TEST SORUSUNU DOĞRULAMA
# ============================================================

# Yeni inference test sorusunu belirliyoruz.
test_question = "Yapay zekâ sistemlerinde model değerlendirmesi neden önemlidir?"

# Soru metinlerini karşılaştırılabilir biçime getiriyoruz.
dataset_questions_normalized = {
    question.strip().casefold()
    for question in dataset_questions
}

# Yeni sorunun veri setinde birebir bulunup bulunmadığını kontrol ediyoruz.
question_is_in_dataset = (
    test_question.strip().casefold()
    in dataset_questions_normalized
)

print("Test sorusu:", test_question)
print("Veri setinde birebir mevcut mu?:", question_is_in_dataset)

# Soru veri setinde varsa inference aşamasına geçmeden duruyoruz.
assert not question_is_in_dataset, (
    "Bu soru veri setinde mevcut. Lütfen farklı bir soru seçin."
)

print("Kontrol başarılı: Soru veri setinde birebir bulunmuyor.")
```

```python

# ============================================================
# 13.1. ADIM: EĞİTİLMİŞ LORA MODELİYLE INFERENCE
# ============================================================

import torch

# Modeli değerlendirme moduna alıyoruz.
model.eval()

# Veri setinde birebir bulunmadığını doğruladığımız soruyu kullanıyoruz.
test_question = "Yapay zekâ sistemlerinde model değerlendirmesi neden önemlidir?"

# Eğitim sırasında kullandığımız soru-cevap biçiminde bir girdi oluşturuyoruz.
prompt = (
    "### Soru:\n"
    + test_question
    + "\n\n### Cevap:\n"
)

# Soruyu token'lara dönüştürüp modelin bulunduğu cihaza aktarıyoruz.
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Gradyan hesaplamadan modelin cevap üretmesini sağlıyoruz.
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id
    )

# Girdi olarak verilen soru kısmını ayırıp yalnızca yeni üretilen token'ları alıyoruz.
new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

# Token'ları okunabilir Türkçe metne dönüştürüyoruz.
generated_answer = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print("Test sorusu:", test_question)
print("\nModelin ürettiği cevap:")
print(generated_answer)
```
